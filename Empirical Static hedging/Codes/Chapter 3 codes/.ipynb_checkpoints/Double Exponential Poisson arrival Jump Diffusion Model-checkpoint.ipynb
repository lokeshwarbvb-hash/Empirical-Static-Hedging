```python
#Function to generate covariance matrix
def generate_covariance_from_correlation(cor_mat, vol_list, dt):
    vol_diag_mat = np.diag(vol_list)
    cov_mat = np.dot(np.dot(vol_diag_mat, cor_mat), vol_diag_mat) * dt
    return cov_mat

#Generate Simulations of stocks - GBM
def multi_variate_gbm_simulation(no_of_paths, no_of_exercise_days, exercise_days, no_of_assets,
                                 curr_stock_price, r, vol_list, cov_mat, t):
    zero_mean = np.zeros(no_of_assets)
    dt = float(t / no_of_exercise_days)
    dw_matx = np.random.multivariate_normal(zero_mean, cov_mat, (int(no_of_paths / 2), no_of_exercise_days))
    dw_mat = np.concatenate((dw_matx, -dw_matx), axis=0)
    dw_mat = dw_mat.reshape(no_of_paths, no_of_exercise_days)

    sim_ln_stock_mat = np.zeros((no_of_paths, no_of_exercise_days + 1))
    sim_ln_stock_mat[:, 0] = np.log(curr_stock_price)
    base_drift = (r - 0.5 * np.square(vol_list[0])) * dt

    for day in range(1, no_of_exercise_days + 1):
        curr_drift = sim_ln_stock_mat[:, day - 1] + base_drift
        sim_ln_stock_mat[:, day] = curr_drift + dw_mat[:, day - 1]

    sim_stock_mat = np.exp(sim_ln_stock_mat)
    return sim_stock_mat
```

```python
#Add jumps to GBM

def add_jumps_to_GBM(gbm_sim_mat, no_of_exercise_days, lamda, p, eta_1, eta_2 t):
    dt = float(t / no_of_exercise_days)
    q = 1-p
    epsilon = (p*eta_1 / (eta_1 - 1)) + (q*eta_1 / (eta_2 + 1)) - 1
    const_term = - lamda * epsilon * dt
    
    no_of_rows, no_of_cols = gbm_sim_mat.shape
    poisson_mat = np.zeros((no_of_rows, no_of_cols))
    poisson_mat[,1:] = np.random.poisson(lamda, no_of_rows * (no_of_cols-1)).reshape(no_of_rows, no_of_cols-1)
    
    return 0
    
```

```python
#European Call option price under Kou's Double Exponential Distribution Model
#Parameters: r, vol, lamda, eta_1, eta_2, p

def nCr(n, r):
    return (math.factorial(n) / (math.factorial(r) * math.factorial(n-r)))

def Hh_func(n, x):
    if (n == -1):
        return (np.sqrt(math.pi) * norm.pdf(x))
    elif (n == 0):
        return (np.sqrt(math.pi) * norm.cdf(-x))
    else:
        return (Hh_func(n-2, x) - x * Hh_func(n-1, x)) / n
    
def poisson_prob(lamda, n, T):
    return (np.exp(-lamda * T) * ((lamda * T)**n)) / math.factorial(n)

def I_func(n, c, a, b, d):
    b_by_a = np.array([(b/a)**(n-i) for i in range(0, n+1)])
    Hh_array = np.array([Hh_func(i, b*c-d) for i in range(0, n+1)])
    
    """ 
    I think the first if condition is the only possible case 
    for the double exponential Jump diffusion process """
    
    if (b<0 and a < 0):
        return - (np.exp(a*c) / a) * np.dot(b_by_a, Hh_array) \
               - (b/a)**(n+1) * (np.sqrt(2* math.pi) / b) * np.exp(a*d/b + (a**2)/(2 * (b**2))) * norm.cdf(-b*c + d + (a/b))
    elif (b>0 and a != 0):
        return - (np.exp(a*c) / a) * np.dot(b_by_a, Hh_array) \
               + (b/a)**(n+1) * (np.sqrt(2* math.pi) / b) * np.exp(a*d/b + (a**2)/(2 * (b**2))) * norm.cdf(-b*c + d + (a/b))
    elif (b>0 and a==0):
        return (1/b) * Hh_func(n+1, b*c - d)
    elif (b==0 and a < 0):
        return Hh_func(-d) * np.exp(a*c)
    else: 
        return float('inf')

def P_n_k(n, k, p, eta_1, eta_2):
    q = 1-p
    if (n != k):
        P_n_k_array = np.array([nCr(n-k-1, i-k) * nCr(n, i) \
                                * ((eta_1/(eta_1 + eta_2))**(i-k)) \
                                * ((eta_2/(eta_1 + eta_2))**(n-i)) \
                                * (p**(i)) *(q**(n-i)) for i in range(k, n)])
        return np.sum(P_n_k_array)
    else:
        return (p**n)

def Q_n_k(n, k, p, eta_1, eta_2):
    q = 1-p
    if (n != k):
        Q_n_k_array = np.array([nCr(n-k-1, i-k) * nCr(n, i) \
                                * ((eta_1/(eta_1 + eta_2))**(n-i)) \
                                * ((eta_2/(eta_1 + eta_2))**(i-k)) \
                                * (p**(n-i)) *(q**(i)) for i in range(k, n)])
        return np.sum(Q_n_k_array)
    else:
        return ((1-p)**n)
    
    """ 
    P(Z(T) >= a) is a sum of infinites series - but we have restricted the sum with max n as 15.
    This is in line with Kou's paper for the double exponential Jump diffusion process """
    
def prob_ZofT_gte_a(r, vol, lamda, p, eta_1, eta_2, a, T, n_max=15):
    
    P_n_k_mat = np.zeros((n_max, n_max)).reshape(n_max, n_max)
    Q_n_k_mat = np.zeros((n_max, n_max)).reshape(n_max, n_max)
    
    for i in range(0, n_max):
        for j in range(0, n_max):
            if (i>=j):
                P_n_k_mat[i, j] = P_n_k(i+1, j+1, p, eta_1, eta_2)
                Q_n_k_mat[i, j] = Q_n_k(i+1, j+1, p, eta_1, eta_2)
    
    for i in range(0, n_max):
        I_kminus_P_array = np.array([ I_func(i, a - r*T, - eta_1, -1/(vol * np.sqrt(T)), -vol * eta_1 * np.sqrt(T)) \
                                   for i in range(0, n_max)]).reshape(n_max, 1)
        I_kminus_Q_array = np.array([ I_func(i, a - r*T, eta_2, 1/(vol * np.sqrt(T)), -vol * eta_2 * np.sqrt(T)) \
                                   for i in range(0, n_max)]).reshape(n_max, 1)
        poisson_prob_array = np.array([poisson_prob(lamda, i, T) for i in range(1, n_max+1)]).reshape(1, n_max)
    
    const1 = np.exp(((vol * eta_1)**2) * (T/2)) / (vol * np.sqrt(2 * math.pi * T))
    const2 = np.exp(((vol * eta_2)**2) * (T/2)) / (vol * np.sqrt(2 * math.pi * T))
    const3 = poisson_prob(lamda, 0, T) * norm.cdf(-((a - r*t)/(vol * np.sqrt(t))))
    
    prob_for_ZofT_gte_a = const1 * np.matmul(poisson_prob_array, np.matmul(P_n_k_mat, I_kminus_P_array)) \
                        + const2 * np.matmul(poisson_prob_array, np.matmul(Q_n_k_mat, I_kminus_Q_array)) \
                        + const3
    
    return prob_for_ZofT_gte_a

def call_price_by_jump_model(S_0, K, r, vol, lamda, p, eta_1, eta_2, T, n_max=15):
    a = np.log(K/S_0)
    q = 1-p
    epsilon = (p*eta_1 / (eta_1 - 1)) + (q*eta_1 / (eta_2 + 1)) - 1
    p_new = (p / (1 + epsilon)) * (eta_1 /(eta_1 - 1))
    eta_1_new = eta_1 - 1
    eta_2_new = eta_2 + 1
    lamda_new = lamda * (1 + epsilon)
    price = S_0 * prob_ZofT_gte_a(r + (0.5 * vol**2) - (lamda * epsilon), vol, lamda_new, p_new, eta_1_new, eta_2_new, a, T, n_max) \
            - K * np.exp(r*T) * prob_ZofT_gte_a(r - (0.5 * vol**2) - (lamda * epsilon), vol, lamda, p, eta_1, eta_2, a, T, n_max)
    return price
```

```python
def BSM_call(s_t, k, r, vol, dt, w=np.array([1]).reshape(-1,1)):
    d_1_c = (1 / (vol * (dt ** 0.5))) 
    d1_log =  (np.log((w * s_t)/(k)) + ((r + (vol ** 2) / 2) * dt))
    d_1 = d_1_c * d1_log
    d_2 = d_1 - vol * (dt ** 0.5)
    return (np.multiply(norm.cdf(d_1) * w,  s_t) - (norm.cdf(d_2) * k * np.exp(-r * dt)))[0][0]

print(BSM_call(100, 98, 0.05, 0.16, 0.5))
```
