```python
# Required Python packages
from datetime import date, datetime, timedelta
import calendar
import pandas as pd
import numpy as np
import itertools
from scipy.stats import norm
import copy
import os

#Import pricing functions
from ipynb.fs.full.user_defined_vik_functions import BSM_call, BSM_put

#Import Date opertative user defined functions
from ipynb.fs.full.user_defined_vik_functions import get_all_monthly_option_expiries, \
                                                     find_last_thurs_date_of_month, \
                                                     prev_workday_if_holiday, prev_weekday, find_wkly_expries,\
                                                     date_of_prev_thurs

#Import IV functions
from ipynb.fs.full.A1_BS_Generate_daily_Implied_Vol_surface import gen_brents_iv, error_func, \
                                                                   product_other_elem, \
                                                                   inverse_quadratic_iv, \
                                                                   secant_method_iv, \
                                                                   bisection_method_iv, swap

# Import dataframe naming functions 
from ipynb.fs.full.user_defined_vik_functions import get_mthly_df_name_from_expiry

#Import data loading functions
from ipynb.fs.full.user_defined_vik_functions import load_all_mthly_data

#Import Risk-free interest rate generation functions
from ipynb.fs.full.user_defined_vik_functions import generate_weekly_ir

#Import implied volatility processing functions
from ipynb.fs.full.user_defined_vik_functions import generate_weekly_iv, gen_interpolated_iv

#Path Settings
source_path = "/home/jupyter-partha/Vikranth - Chapter 2/"
input_sub_path = "Input Data/mkt_data_covid_region/"
output_sub_path = "Output Data/"
input_data_path = source_path + input_sub_path
output_data_path = source_path + output_sub_path


# Stock Index of Interest
stock_ident = "BANKNIFTY"

# The scope of this code is to hedge one option 
# and scope will be extended to a portfolio
no_of_assets = 1
cor_mat = [[1]]

prod_moneyness = "ATM"

#Product type to hedge: either "CE" or "PE"
prod_type = "CE"

# Periods of interest will be a dictionary
#Key is the year, value is a list of months 1-12, 1- Jan, 2 - Feb,...12 - Dec
# For E.g., periods_of_interest = {2020: [3], 2019: [11, 12]}
periods_of_interest = {2020: [3]}

#List of holidays
holidays_list = [date(2020, 3, 10)]
```

```python
each_month = date(2020,3,26)

mthly_expiries_list = get_all_monthly_option_expiries(periods_of_interest, holidays_list)
dict_wkly_expiries_each_month = find_wkly_expries(mthly_expiries_list, holidays_list)
mthly_mkt_data = load_all_mthly_data(mthly_expiries_list, input_data_path, prod_type_lists=["FUT", "CE", "PE"], stock_ident=stock_ident)
dict_wkly_ir = generate_weekly_ir(dict_wkly_expiries_each_month, mthly_mkt_data, stock_ident)

#Pricer
#BSM_call Parameters -> (s_t, k, r, vol, dt):
#gen_interpolated_iv(df_iv_surface, vol_date_str, moneyness)
df_iv_surface = pd.read_csv(output_data_path + "A1_Daily_vol_smile_CE_" + each_month.strftime("%d-%b-%Y") + ".csv")
Strike = 30200.00

# 5-March
Spot = 28815.35
moneyness = float(Spot/Strike)
week = 0
r_5Mar = np.array([dict_wkly_ir[each_month.strftime("%d-%b-%Y")][week]])
vol_date_str = date(2020,2,27).strftime("%d-%b-%Y")
iv_5Mar_ATM = gen_interpolated_iv(df_iv_surface, vol_date_str, 1)
iv_5Mar_SMILE = gen_interpolated_iv(df_iv_surface, vol_date_str, moneyness)
PV_5Mar_ATM = BSM_call(Spot, Strike, r_5Mar, iv_5Mar_ATM, 0.05753)
PV_5Mar_SMILE = BSM_call(Spot, Strike, r_5Mar, iv_5Mar_SMILE, 0.05753)

# # 12-March
Spot = 23971.15
moneyness = float(Spot/Strike)
week = 1
r_12Mar = np.array([dict_wkly_ir[each_month.strftime("%d-%b-%Y")][week]])
vol_date_str = date(2020,3,5).strftime("%d-%b-%Y")
iv_12Mar_ATM = gen_interpolated_iv(df_iv_surface, vol_date_str, 1)
iv_12Mar_SMILE = gen_interpolated_iv(df_iv_surface, vol_date_str, moneyness)
PV_12Mar_ATM = BSM_call(Spot, Strike, r_12Mar, iv_12Mar_ATM, 0.03836)
PV_12Mar_SMILE = BSM_call(Spot, Strike, r_12Mar, iv_12Mar_SMILE, 0.03836)
print(BSM_call(Spot, Strike, -0.137510499134541, 0.569153044997332, 0.03836))
print(BSM_call(Spot, Strike, r_12Mar, 0.569153044997332, 0.03836))
print(BSM_call(Spot, Strike, -0.137510499134541, iv_12Mar_SMILE, 0.03836))

# 19-March
Spot = 20083.50
moneyness = float(Spot/Strike)
week = 2
r_19Mar = np.array([dict_wkly_ir[each_month.strftime("%d-%b-%Y")][week]])
vol_date_str = date(2020,3,12).strftime("%d-%b-%Y")
iv_19Mar_ATM = gen_interpolated_iv(df_iv_surface, vol_date_str, 1)
iv_19Mar_SMILE = gen_interpolated_iv(df_iv_surface, vol_date_str, moneyness)
PV_19Mar_ATM = BSM_call(Spot, Strike, r_19Mar, iv_19Mar_ATM, 0.01918)
PV_19Mar_SMILE = BSM_call(Spot, Strike, r_19Mar, iv_19Mar_SMILE, 0.01918)
print(BSM_call(Spot, Strike, -0.142992567816506, 1.2628228669735, 0.01918))
print(BSM_call(Spot, Strike, r_19Mar, 1.2628228669735, 0.01918))
print(BSM_call(Spot, Strike, -0.142992567816506, iv_19Mar_SMILE, 0.01918))

```
