```python
# Required Python packages
from datetime import date, datetime, timedelta
import calendar
import pandas as pd
import numpy as np
import itertools
from scipy.stats import norm
import copy
import os

#Import pricing functions
from ipynb.fs.full.user_defined_vik_functions import BSM_call, BSM_put

#Import IV functions
from ipynb.fs.full.A1_BS_Generate_daily_Implied_Vol_surface import gen_brents_iv, error_func, \
                                                                   product_other_elem, \
                                                                   inverse_quadratic_iv, \
                                                                   secant_method_iv, \
                                                                   bisection_method_iv, swap
#Path Settings
source_path = "/home/jupyter-partha/Vikranth - Chapter 2/"
input_sub_path = "Input Data/"
output_sub_path = "Output Data/"
input_data_path = source_path + input_sub_path
output_data_path = source_path + output_sub_path
```

```python
# Input NSE data

nse_data = pd.read_csv(input_data_path + "mkt_data_mibor_iv_22Feb.csv")
nse_data= nse_data.assign(BSM_Call=lambda x: BSM_call(x['Spot'], x['Strike'], x['r'], x['Call IV']/100, x['T']))
nse_data= nse_data.assign(BSM_Put=lambda x: BSM_put(x['Spot'], x['Strike'], x['r'], x['Put IV']/100, x['T']))
# nse_data= nse_data.assign(call_brents_iv = lambda x: gen_brents_iv(x['Call MV'], x['Spot'], x['Strike'], x['r'], x['T'], option_type="CE", iv_list_initial=[0.0000000001, 50, 50]))
# nse_data= nse_data.assign(BSM_Put=lambda x: BSM_put(x['Spot'], x['Strike'], x['r'], x['Put IV']/100, x['T']))                                                                       
# nse_data.dtypes
```

```python
call_iv_list = []
put_iv_list = []

for i in range(0, len(nse_data)):
#     (mkt_price, s_t, k, r, dt, option_type, iv_list)
    iv_call = gen_brents_iv(nse_data.iloc[i,4], nse_data.iloc[i,0], nse_data.iloc[i,5], \
                       nse_data.iloc[i,1], nse_data.iloc[i,2], \
                       option_type="CE", iv_list_initial=[0.0000000001, 50, 50])

    iv_put = gen_brents_iv(nse_data.iloc[i,6], nse_data.iloc[i,0], nse_data.iloc[i,5], \
                       nse_data.iloc[i,1], nse_data.iloc[i,2], \
                       option_type="PE", iv_list_initial=[0.0000000001, 50, 50])
    
    call_iv_list.append(iv_call)
    put_iv_list.append(iv_put)

nse_data['Brents Call IV'] = call_iv_list
nse_data['Brents Put IV'] = put_iv_list
```

```python
print(nse_data)
nse_data.to_csv(output_data_path + "T1_" + "mibor_iv_checks" + ".csv", index = False)
```
