```python
#Lognormal simulations

from scipy.stats import lognorm
import numpy as np
import matplotlib.pyplot as plt

#scale =exp(mu), where mu is the mean of normal random variable 
#when you take on the lognormal variable
#s= sigma of normal variable
s_0 = 100
vol = 0.8
r=0.07
t = 1
# Normal variable: log s_t = log s_0 + [r - (sigma^2)/2] * t + sigma * t * Z(0,1)
mu = np.log(100) + (r - (vol**2)/2) * t
std_dev = vol * np.sqrt(t)
sample = lognorm.rvs(s=std_dev, loc=0, scale=np.exp(mu), size=100000)
# plt.hist(sample, normed=True, bins=1000, alpha=0.6, color='c', ec='c')
bins = range(0, 1000, 20)
plt.hist(sample, bins = bins)
plt.show()

print("1pct, 5pct, 10pct, 25pct, 50pct, 75pct, 90pct, 95pct, 99pct: ")
print(np.percentile(sample, [1, 5, 10, 25, 50, 75, 90, 95, 99]))
print(np.mean(sample))
```

```python
lamda_exp = 0.5
exp_sample = np.random.exponential(scale = (1/lamda_exp), size=100)
bins = range(0, 100, 1) 
bins = [bin/10 for bin in bins]
plt.hist(exp_sample, bins = bins)
plt.show()
print("1pct, 5pct, 10pct, 25pct, 50pct, 75pct, 90pct, 95pct, 99pct: ")
print(np.percentile(exp_sample, [1, 5, 10, 25, 50, 75, 90, 95, 99]))
print(np.mean(exp_sample))
```
