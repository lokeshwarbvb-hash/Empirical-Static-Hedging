```python
###################
# Python Packages #
##################

import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as st
```

```python
def COSDensity(cf, X, N, a, b):
    i = np.complex(0.0, 1.0)
    k = np.linspace(0, N-1, N)
    u = np.zeros([1, N])
    u = (k * np.pi) / (b-a)

    F_k = 2.0 / (b-a) * np.real(cf(u) * np.exp(-i * u * a))
    F_k[0] = F_k[0] * 0.5

    f_X = np.matmul(F_k, np.cos(np.outer(u, x-a)))
    
    return f_X
    
def mainCalculation():
    i = np.complex(0.0, 1.0)
    a = -10
    b = 10
    N = [2**i for i in range(2, 7, 1)]
    
    mu = 0
    sigma = 1
    
    # Characteristic Function of Normal Random Variable (X)
    # E[e^{iux}] = exp(i*mu*u - 0.5 * sigma^2 * u^2)
    cF = lambda u: np.exp(i*mu*u - 0.5 * np.power(sigma, 2.0) * np.power(u, 2.0))
    x = np.linspace(-5,5,1000)
    f_XExact = st.norm.pdf(x,mu,sigma)
    
    plt.figure()
    plt.grid()
    plt.xlabel("X")
    plt.ylabel("$f_X(x)$")
    plt.plot(x, f_XExact)
    
    for n in N:
        f_X = COSDensity(cF, x, n, a, b)
        error = np.max(np.abs(f_X - f_XExact))
        print("For {0} expansion terms, the error is {1}".format(n, error))
        plt.plot(x, f_X)
        
    plt.legend(["Exact"] + ["N=%.0f"%N[i] for i in range(0, len(N))])

mainCalculation()
       
```
